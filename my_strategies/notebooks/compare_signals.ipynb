```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path("../data/signals")

SIGNAL_FILE = DATA_DIR / "signals_2024-12-01_2026-02-21_20260221_125721.parquet"
SIGNAL_V2_FILE = DATA_DIR / "signals_v2_2024-12-01_2026-02-24_20260224_041123.parquet"

df_old = pd.read_parquet(SIGNAL_FILE)
df_new = pd.read_parquet(SIGNAL_V2_FILE)

print(f"Old signal shape: {df_old.shape}")
print(f"New signal shape: {df_new.shape}")
```

```python
old_symbols = set(df_old["symbol"].unique())
new_symbols = set(df_new["symbol"].unique())

print(f"Old unique symbols: {len(old_symbols)}")
print(f"New unique symbols: {len(new_symbols)}")
print(f"\nExact match: {old_symbols == new_symbols}")
```

```python
only_in_old = sorted(old_symbols - new_symbols)
only_in_new = sorted(new_symbols - old_symbols)

print(f"Symbols only in old ({len(only_in_old)}): {only_in_old}")
print(f"Symbols only in new ({len(only_in_new)}): {only_in_new}")
```
